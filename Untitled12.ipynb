```python
import pandas as pd
# Criando a nossa tabela de discursos e sentimentos
dados = pd.DataFrame({
    'Texto': [
        "Estou muito ansioso e com o coração acelerado hoje.",
        "Que dia maravilhoso! Estou muito feliz.",
        "Não aguento mais esse estresse e cobrança na minha cabeça.",
        "Amei passar o tempo com os meus amigos, estou alegre.",
        "Sinto uma paranoia constante, parece que estão me perseguindo.",
        "Tudo parece dar errado, sinto uma tristeza profunda."
    ],
    'Sentimento': ['Ansiedade', 'Feliz', 'Estresse', 'Feliz', 'Paranoia', 'Tristeza']
})
# Mostra a tabela na tela
dados
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB

# 1. Transformando as palavras em números (Vetorização TF-IDF)
transformador = TfidfVectorizer()
numeros = transformador.fit_transform(dados['Texto'])

# 2. Criando o modelo de IA (Naive Bayes)
inteligencia_artificial = MultinomialNB()

# 3. Treinando a IA com os nossos dados
inteligencia_artificial.fit(numeros, dados['Sentimento'])

print("Pronto! A Inteligência Artificial aprendeu com sucesso.")
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# 1. Fazendo a IA tentar adivinhar os sentimentos com base no que ela aprendeu
previsoes = inteligencia_artificial.predict(numeros)

# 2. Criando a Matriz de Confusão
matriz = confusion_matrix(dados['Sentimento'], previsoes, labels=inteligencia_artificial.classes_)

# 3. Desenhando o gráfico na tela
plt.figure(figsize=(6, 6))
grafico = ConfusionMatrixDisplay(confusion_matrix=matriz, display_labels=inteligencia_artificial.classes_)
grafico.plot(cmap=plt.cm.Blues)
plt.title("Matriz de Confusão - Trabalho do Módulo 2")
plt.show()
```

```python
# ==============================================================================
# PÓS-GRADUAÇÃO EM IA APLICADA À SAÚDE - SANTA CASA DE MISERICÓRDIA
# MÓDULO 2: IA E CIÊNCIA DE DADOS
# PROJETO: ANTROPOLOGIA E SAÚDE MENTAL
# ==============================================================================

# ------------------------------------------------------------------------------
# 1. BIBLIOTECAS (As ferramentas que vamos usar)
# ------------------------------------------------------------------------------
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report, accuracy_score

# ------------------------------------------------------------------------------
# BASE DE DADOS (Simulação dos discursos de redes sociais)
# ------------------------------------------------------------------------------
print("--- Carregando a Base de Dados ---")
dados_saude_mental = {
    'Text': [
        "Estou me sentindo muito ansioso, com o coração acelerado e sem forças para sair de casa hoje.",
        "Que dia maravilhoso! Consegui cumprir todas as minhas metas no trabalho e estou muito feliz.",
        "Não aguento mais essa cobrança e pressão, sinto um estresse absurdo e esgotamento mental na cabeça.",
        "Amei passar o tempo com os meus amigos hoje no parque, me sinto totalmente renovado e alegre.",
        "Me sinto perseguido nas redes sociais, sinto muito medo e uma paranoia constante do que falam de mim.",
        "Praticar exercícios físicos pela manhã e meditar me traz uma paz de espírito incrível.",
        "Tudo parece dar errado na minha vida, sinto uma tristeza profunda e um vazio que não passa nunca.",
        "Excelente leitura hoje sobre antropologia, comportamento humano e evolução das sociedades.",
        "Essa polarização política extrema no Twitter me deixa extremamente paranóico, irritado e estressado.",
        "Estou muito otimista e esperançoso com os resultados do meu novo tratamento e acompanhamento de saúde."
    ],
    'Sentiment': ['Ansiedade', 'Feliz', 'Estresse', 'Feliz', 'Paranoia', 'Feliz', 'Tristeza', 'Feliz', 'Estresse', 'Feliz']
}

df = pd.DataFrame(dados_saude_mental)
print(f"Total de registros carregados: {len(df)}\n")

# ------------------------------------------------------------------------------
# LIMPEZA DOS TEXTOS (Removendo pontuação e padronizando letras minúsculas)
# ------------------------------------------------------------------------------
print("--- Executando a Limpeza dos Textos ---")
def limpar_texto(texto):
    texto = texto.lower() # Transforma tudo em minúsculo
    texto = re.sub(r'[^\w\s]', '', texto) # Remove pontos, vírgulas e exclamações
    return texto

df['Texto_Limpo'] = df['Text'].apply(limpar_texto)
print("Exemplo de texto limpo:")
print(f"Original: {df['Text'][0]}")
print(f"Limpo:    {df['Texto_Limpo'][0]}\n")

# ------------------------------------------------------------------------------
# ANÁLISE EXPLORATÓRIA (Entendendo a distribuição das classes)
# ------------------------------------------------------------------------------
print("--- Análise Exploratória (Distribuição de Sentimentos) ---")
distribuicao = df['Sentiment'].value_counts()
print(distribuicao)

# Gerando um gráfico simples de barras para a análise exploratória
plt.figure(figsize=(6, 4))
sns.barplot(x=distribuicao.index, y=distribuicao.values, palette='Blues_r')
plt.title('Quantidade de Discursos por Categoria de Saúde Mental')
plt.xlabel('Sentimento / Condição')
plt.ylabel('Quantidade')
plt.show()
print("\n")

# ------------------------------------------------------------------------------
# VETORIZAÇÃO COM TF-IDF (Transformando palavras em pesos numéricos relevantes)
# ------------------------------------------------------------------------------
print("--- Vetorização com TF-IDF ---")
# O TF-IDF calcula a importância de cada palavra no discurso
vetorizador_tfidf = TfidfVectorizer()
X = vetorizador_tfidf.fit_transform(df['Texto_Limpo'])
y = df['Sentiment']
print(f"Formato da matriz gerada pelo TF-IDF (Linhas, Palavras únicas): {X.shape}\n")

# ------------------------------------------------------------------------------
# MODELO DE CLASSIFICAÇÃO (Treinando a IA)
# ------------------------------------------------------------------------------
print("--- Treinando o Modelo (Naive Bayes) ---")
# Dividimos os dados: 70% para a IA aprender (Treino) e 30% para testarmos (Teste)
X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, test_size=0.3, random_state=42)

modelo_nb = MultinomialNB()
modelo_nb.fit(X_treino, y_treino)
print("Modelo Naive Bayes treinado com sucesso!\n")

# ------------------------------------------------------------------------------
# AVALIAÇÃO (Matriz de Confusão e Métrica de Acerto)
# ------------------------------------------------------------------------------
print("--- Avaliação do Modelo ---")
y_previsao = modelo_nb.predict(X_teste)

# Gerando e exibindo a Matriz de Confusão
matriz = confusion_matrix(y_teste, y_previsao, labels=modelo_nb.classes_)

plt.figure(figsize=(6, 6))
disp = ConfusionMatrixDisplay(confusion_matrix=matriz, display_labels=modelo_nb.classes_)
disp.plot(cmap=plt.cm.Blues)
plt.title("Matriz de Confusão - Módulo 2 (TF-IDF)")
plt.show()

acuracia = accuracy_score(y_teste, y_previsao)
print(f"Acurácia Geral do Modelo: {acuracia * 100:.2f}%\n")
print("Relatório de Classificação:")
print(classification_report(y_teste, y_previsao, labels=modelo_nb.classes_, zero_division=0))
```
