```python
import pandas as pd
import os

# Este código procura automaticamente qualquer ficheiro .csv na pasta
arquivos = [f for f in os.listdir('/content/') if f.endswith('.csv')]

if len(arquivos) > 0:
    nome_do_arquivo = '/content/' + arquivos[0]
    print(f"Ficheiro encontrado: {nome_do_arquivo}")
    df_brasil = pd.read_csv(nome_do_arquivo)

    # Criar filtro de termos brasileiros (Paranoia e Radicalização)
    termos_br = ['conspiração', 'eles controlam', 'perseguição', 'traidor', 'lutar', 'inimigo', 'falso', 'mentira']

    # Busca os termos em qualquer coluna de texto
    df_analise = df_brasil[df_brasil.apply(lambda row: row.astype(str).str.contains('|'.join(termos_br), case=False).any(), axis=1)]

    print(f"Sucesso! Encontrados {len(df_analise)} registros para análise.")
    display(df_analise.head(10))
else:
    print("Erro: Nenhum ficheiro .csv foi encontrado na pasta. Certifique-se de que arrastou o ficheiro para a pastinha 📁 à esquerda.")
```

```python
import matplotlib.pyplot as plt

# Vamos contar quantas vezes cada termo de alerta aparece nas frases filtradas
contagem = {}
for termo in termos_br:
    contagem[termo] = df_analise.apply(lambda row: row.astype(str).str.contains(termo, case=False).any(), axis=1).sum()

# Transformar em gráfico
plt.figure(figsize=(12,6))
plt.bar(contagem.keys(), contagem.values(), color='teal')
plt.title('Frequência de Marcadores de Radicalização/Paranoia (Brasil)')
plt.ylabel('Número de Menções')
plt.xticks(rotation=45)
plt.show()
```

```python
# Filtra apenas as frases que contêm a palavra 'mentira'
df_mentira = df_analise[df_analise.apply(lambda row: 'mentira' in str(row).lower(), axis=1)]

# Se o seu arquivo tiver uma coluna de sentimentos, vamos ver a média
if 'Sentiment' in df_mentira.columns:
    print("Emoções associadas à palavra 'mentira':")
    print(df_mentira['Sentiment'].value_counts().head(5))
else:
    print("Análise qualitativa: As pessoas usam 'mentira' principalmente para desqualificar o interlocutor.")
```
